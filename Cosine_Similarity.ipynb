# Calculating Cosine Similarity Mathematically

## Aim
To understand and calculate cosine similarity between two vectors mathematically using Python.

## Theory

Cosine similarity measures the angle between two vectors. It is commonly used to determine how similar two pieces of text or documents are.

Formula:

Cosine Similarity = (A · B) / (||A|| × ||B||)

```python
import math

A = [2, 3, 1]
B = [4, 1, 2]

dot_product = sum(a * b for a, b in zip(A, B))

magnitude_A = math.sqrt(sum(a * a for a in A))
magnitude_B = math.sqrt(sum(b * b for b in B))

cosine_similarity = dot_product / (magnitude_A * magnitude_B)

print('Vector A:', A)
print('Vector B:', B)
print('Dot Product:', dot_product)
print('Magnitude of A:', round(magnitude_A, 4))
print('Magnitude of B:', round(magnitude_B, 4))
print('Cosine Similarity:', round(cosine_similarity, 4))
```

## Using NumPy

```python
import numpy as np

A = np.array([2, 3, 1])
B = np.array([4, 1, 2])

similarity = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))

print('Cosine Similarity using NumPy:', round(similarity, 4))
```

## Interpretation

A cosine similarity value closer to 1 indicates that the vectors are more similar, while a value closer to 0 indicates that they are less similar.

## Conclusion

Cosine similarity was calculated manually using the dot product and vector magnitudes. The same result was verified using NumPy.
