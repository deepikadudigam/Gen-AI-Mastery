# Introduction to Vector Databases — ChromaDB

## Aim
To understand the basic concept of vector databases and store text embeddings using ChromaDB.

## Introduction

A vector database stores numerical vector representations of data and allows efficient similarity-based searching. ChromaDB is an open-source vector database commonly used in AI and RAG applications.

```python
!pip install -q chromadb
```

```python
import chromadb

client = chromadb.Client()

collection = client.create_collection(name='genai_documents')

print('ChromaDB collection created successfully.')
```

## Adding Documents

Documents can be stored along with unique IDs. ChromaDB automatically handles the vector representation when embeddings are configured.

```python
documents = [
    'Generative AI can create text, images and code.',
    'Machine learning allows computers to learn from data.',
    'Natural Language Processing works with human language.',
    'Vector databases are useful for semantic search.'
]

ids = ['doc1', 'doc2', 'doc3', 'doc4']

collection.add(
    documents=documents,
    ids=ids
)

print('Documents added successfully.')
```

```python
print('Number of documents stored:', collection.count())
```

## Similarity Search

```python
query = 'What is semantic search?'

results = collection.query(
    query_texts=[query],
    n_results=2
)

print('Query:', query)
print('\nMost relevant documents:')

for doc in results['documents'][0]:
    print('-', doc)
```

## Conclusion

ChromaDB provides a convenient way to store documents and perform similarity-based retrieval. Vector databases form an important component of modern Generative AI and RAG systems.
