# Word Embeddings

This notebook demonstrates basic word representation techniques using One-Hot Encoding and Word2Vec.

## 1. One-Hot Encoding

One-Hot Encoding represents each word as a binary vector. Each word is assigned a unique position in the vocabulary.

```python
from sklearn.preprocessing import OneHotEncoder

words = [['python'], ['machine'], ['learning'], ['analytics']]

encoder = OneHotEncoder(sparse_output=False)
encoded_words = encoder.fit_transform(words)

print('Vocabulary:', encoder.categories_[0])
print('\nOne-Hot Encoded Vectors:')
print(encoded_words)
```

## 2. Word2Vec Embeddings

Word2Vec learns numerical vector representations of words based on their context in sentences.

```python
!pip install -q gensim
```

```python
from gensim.models import Word2Vec

sentences = [
    ['data', 'science', 'uses', 'python'],
    ['data', 'analysis', 'uses', 'python'],
    ['machine', 'learning', 'uses', 'data'],
    ['deep', 'learning', 'uses', 'data'],
    ['python', 'helps', 'with', 'data', 'analysis']
]

model = Word2Vec(
    sentences,
    vector_size=20,
    window=2,
    min_count=1,
    workers=1,
    seed=42
)

print('Word2Vec model created successfully.')
```

```python
word = 'python'

print('Embedding vector for:', word)
print(model.wv[word])

print('\nWords similar to python:')
print(model.wv.most_similar(word, topn=3))
```

## Conclusion

One-Hot Encoding represents words using sparse binary vectors, while Word2Vec learns dense numerical representations based on word context. Word2Vec can capture relationships between words and is widely used in Natural Language Processing.
