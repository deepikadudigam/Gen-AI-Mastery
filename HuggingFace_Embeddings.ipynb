# Hugging Face Embeddings — Scratch to Intermediate

## Aim
To understand how text can be converted into numerical vector representations using a Hugging Face sentence embedding model.

## Introduction

Text embeddings convert sentences or words into numerical vectors. These vectors allow machines to compare the meaning and similarity of text.

```python
!pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

print('Hugging Face embedding model loaded successfully.')
```

```python
sentences = [
    'I am learning Generative AI',
    'Generative AI is an exciting technology',
    'I enjoy studying data analytics',
    'The weather is pleasant today'
]

embeddings = model.encode(sentences)

print('Number of sentences:', len(sentences))
print('Embedding dimension:', embeddings.shape[1])

print('\nFirst sentence:')
print(sentences[0])

print('\nFirst sentence embedding:')
print(embeddings[0])
```

## Comparing Text Embeddings

Cosine similarity can be used to determine how semantically similar two sentence embeddings are.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(embeddings)

print('Cosine Similarity Matrix:\n')
print(similarity_matrix)
```

```python
print('Similarity between sentence 1 and sentence 2:')
print(round(similarity_matrix[0][1], 4))

print('\nSimilarity between sentence 1 and sentence 4:')
print(round(similarity_matrix[0][3], 4))
```

## Conclusion

Hugging Face sentence embedding models convert natural language into dense numerical vectors. These embeddings can be used for semantic similarity, information retrieval, clustering, recommendation systems and Retrieval-Augmented Generation (RAG).
