# RAG Practice

## Aim
To implement a simple Retrieval-Augmented Generation workflow using document chunks, embeddings, similarity search and a language model.

## RAG Workflow

A basic RAG pipeline consists of:

1. Loading documents
2. Splitting documents into chunks
3. Creating embeddings
4. Storing vectors
5. Retrieving relevant chunks
6. Generating an answer using retrieved context

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

print('Embedding model loaded successfully.')
```

## Knowledge Base

```python
documents = [
    'Generative AI creates new content such as text, images, audio and code.',
    'Machine learning enables computers to learn patterns from historical data.',
    'Natural Language Processing allows computers to understand and process human language.',
    'Retrieval-Augmented Generation retrieves relevant information before generating a response.',
    'Vector embeddings represent text as numerical vectors that capture semantic information.',
    'A vector database stores embeddings and supports similarity-based retrieval.'
]

document_embeddings = model.encode(documents)

print('Knowledge base created.')
print('Number of documents:', len(documents))
```

## Query Processing

```python
query = 'How does RAG find useful information?'

query_embedding = model.encode([query])

scores = cosine_similarity(query_embedding, document_embeddings)[0]

ranked_indices = scores.argsort()[::-1]

print('Query:', query)
print('\nRetrieved documents:')

for index in ranked_indices[:3]:
    print(f'\nScore: {scores[index]:.4f}')
    print(documents[index])
```

## Building the Retrieved Context

```python
top_k = 3
retrieved_context = '\n'.join(
    documents[index] for index in ranked_indices[:top_k]
)

print('Retrieved Context:\n')
print(retrieved_context)
```

## Generate a Context-Based Response

In a complete RAG system, the retrieved context would be passed to a generative language model. Here, a simple response template demonstrates the final stage.

```python
response = f'''Based on the retrieved information:

{retrieved_context}

Answer:
RAG first retrieves information that is semantically related to the user's query. The retrieved information can then be provided as context to a language model so that the generated response is grounded in relevant information.'''

print(response)
```

## Conclusion

A basic RAG workflow was implemented by creating document embeddings, comparing a query with the knowledge base using cosine similarity, retrieving the most relevant documents and constructing context for response generation.
