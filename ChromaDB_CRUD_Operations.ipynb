# CRUD Operations in ChromaDB

## Aim
To perform Create, Read, Update and Delete (CRUD) operations on documents stored in a ChromaDB collection.

## Introduction

CRUD stands for Create, Read, Update and Delete. These operations are fundamental for managing data in a vector database.

```python
!pip install -q chromadb
```

```python
import chromadb

client = chromadb.Client()
collection = client.create_collection(name='crud_demo')

print('Collection created successfully.')
```

## 1. CREATE

Add documents and their unique IDs to the collection.

```python
collection.add(
    ids=['1', '2', '3'],
    documents=[
        'Python is popular for data science.',
        'Machine learning finds patterns in data.',
        'Generative AI can create new content.'
    ]
)

print('Documents added.')
print('Document count:', collection.count())
```

## 2. READ

Retrieve documents from the collection using their IDs.

```python
result = collection.get(ids=['1', '2', '3'])

print('Stored Documents:')
for doc_id, document in zip(result['ids'], result['documents']):
    print(doc_id, '->', document)
```

## 3. UPDATE

Modify an existing document using its ID.

```python
collection.update(
    ids=['2'],
    documents=['Machine learning enables systems to learn from examples.']
)

updated = collection.get(ids=['2'])

print('Updated document:')
print(updated['documents'][0])
```

## 4. DELETE

Remove a document from the collection using its ID.

```python
collection.delete(ids=['3'])

print('Document 3 deleted.')
print('Remaining document count:', collection.count())
```

```python
final_data = collection.get()

print('Final documents in collection:')
for doc_id, document in zip(final_data['ids'], final_data['documents']):
    print(doc_id, '->', document)
```

## Conclusion

CRUD operations were successfully demonstrated using ChromaDB. Documents were created, retrieved, updated and deleted from a vector database collection.
