# RAG — Document Loading & Chunking Strategies

## Aim
To understand how documents are loaded and divided into smaller chunks for Retrieval-Augmented Generation (RAG) systems.

## Introduction

Retrieval-Augmented Generation combines information retrieval with a language model. Before documents can be retrieved, large documents are usually divided into smaller chunks.

Chunking improves retrieval by allowing the system to find the most relevant sections of a document.

## Sample Document

```python
document = '''
Generative Artificial Intelligence is a branch of artificial intelligence that can create new content.
It can generate text, images, audio and computer programs.
Large Language Models are trained on large collections of text data.
These models can understand and generate human-like language.
Retrieval-Augmented Generation improves language model responses by providing relevant external information.
The retrieved information is supplied to the model as additional context.
Chunking is an important step because large documents may contain more information than a model can process at once.
''' 

print(document)
```

## Character-Based Chunking

A character-based splitter divides text according to a specified chunk size.

```python
from langchain_text_splitters import CharacterTextSplitter

character_splitter = CharacterTextSplitter(
    separator=' ',
    chunk_size=150,
    chunk_overlap=30
)

character_chunks = character_splitter.split_text(document)

print('Number of character chunks:', len(character_chunks))

for i, chunk in enumerate(character_chunks, 1):
    print(f'\nChunk {i}:')
    print(chunk)
```

## Recursive Character Text Splitting

Recursive splitting attempts to preserve meaningful text boundaries while creating chunks of a suitable size.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=30,
    separators=['\n\n', '\n', '. ', ' ', '']
)

recursive_chunks = recursive_splitter.split_text(document)

print('Number of recursive chunks:', len(recursive_chunks))

for i, chunk in enumerate(recursive_chunks, 1):
    print(f'\nChunk {i}:')
    print(chunk)
```

## Comparing Chunk Sizes

Different chunk sizes can affect retrieval quality. Smaller chunks provide more focused information, while larger chunks preserve more context.

```python
for size in [100, 200, 300]:
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=size,
        chunk_overlap=20
    )
    chunks = splitter.split_text(document)
    print(f'Chunk size {size}: {len(chunks)} chunks')
```

## Conclusion

Documents were successfully divided into smaller chunks using different text-splitting strategies. Choosing an appropriate chunk size and overlap is important for building effective RAG systems.
